```python
import os
import pandas as pd
os.chdir('/Users/emily/PycharmProjects')
from misc import mutation_generation
from misc import rank_and_cluster_mutants
```

```python
starter_seqs= pd.read_csv('/Users/emily/PycharmProjects/PeptideProject/plotting_results/mutation_experimental_sequences.csv')
starter_seqs["plain_seq"] = starter_seqs["Consolidated_Sequences"].str.split((r"\[AMD\]")).str[0]


ncaa = ['NLE','Cha','AIB','ivDde','Chg','GABA']
df = mutation_generation.mutate(starter_seqs["plain_seq"],
            n_mutations=(3, 5),
            ncaa=ncaa,
            cap=100000,
            aa_pool='different_class',
            max_ncaa=2,
            position_window=(0,5))
```

```python
df['mutations_made'] = df.apply(
        lambda r: rank_and_cluster_mutants.get_mutations(r.get("starter_seq"), r.get("mutated_seq"), r.get("positions")),
        axis=1,
    )
```

```python
df['mutations_made'] = df['mutations_made'].astype(str)
```

```python
cys_mutations = df[df['mutations_made'].str.contains("'C',")]
top1500 = cys_mutations.groupby(cys_mutations['starter_seq'], sort=False, group_keys=False).head(1500)
```

```python
top1500.to_csv('cysteine_mutations.csv')
```
